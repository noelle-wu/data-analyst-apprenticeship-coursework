## What is `numpy.datetime64`?

- A class for handling dates and times built into the `numpy` library for scientific computing in Python.

- An alternative to the Python `datetime` class with improved memory efficiency and easy integration with `numpy` functions.

- Use `datetime64` objects in lists and arrays just like number types such as `np.float`. The `64` in the name refers to `numpy` storing each object as `64 bit`  integers.

````python
# We often import numpy library with the alias "np"
import numpy as np
print(np.datetime64)
````

## Creating `np.datetime64` objects.

To declare a `datetime64` object you can either pass an **ISO 8601** formatted string:

````python
from datetime import datetime
import dateutil.tz as tz
import pandas as pd
````

````python
dt1 = np.datetime64("2018-10-03 13:00")
print(dt1)
````

Or initialise using a Python `datetime` object:

````python
pydt1 = datetime(year=2018, month=10, day=3, hour=13)
dt2 = np.datetime64(pydt1)
print(dt2)
````

We declare a Numpy `datetime64` object using a single argument. The `np.datetime64` class takes either a string input or an existing Python `datetime` object as input. This string **must** be formatted in the ISO 8601 specification, where the largest unit (year) comes first and the smallest unit (seconds) comes last. You can also initialise a `numpy` object from an existing Python `datetime` object. ([documentation](https://www.iso.org/iso-8601-date-and-time-format.html))

But you **cannot** declare using individual date and time features as arguments, ie using the same declaration pattern as for Python `datetime`, as can be seen in the example below:

```python
dt3 = np.datetime64(year=2018, month=10, day=3, hour=13)
```

## Comparing `np.datetime64` to `datetime`

Both classes will store the information of a date and time. But the `numpy` version is more **memory efficient**. 

````python
import sys

pdt1 = datetime(year=2018, month=10, day=3, hour=13)
ndt1 = np.datetime64("2018-10-03 13:00:00")

# Compare size in bytes
print(sys.getsizeof(pdt1))
print(sys.getsizeof(ndt1)) 
````

When we previously studied the Python `datetime` package we looked at all the different features of the package and how the class `datetime` fits into the package `datetime`. Now we will look at the `datetime64` class slightly differently, as this is a class that fits into a wider Numpy ecosystem. Not everything we learnt for Python `datetime` will be applicable here, as well as some features for Numpy don't exist for basic Python. The Numpy `datetime64` class is not always a direct replacement and learning about this class doesn't mean you should discard the Python `datetime` class. Each class will be useful for different occasions and you will likely find yourself transferring between the two.  

Now that we can declare Numpy objects, we can compare between Numpy and Python versions of a datetime object. The Numpy library is optimised for memory and efficiency. We can check how large an object is in `bytes` using the `getsizeof` function from the built-in `sys` module. Comparing between two objects, _which hold the same information_ , we see that the Numpy object saves 8 bytes compared to native Python. This might not seem like much now. But what if you start to have massive arrays of thousands of dates?

## What about sequences of datetime objects?

````python
pdt2 = datetime(year=2019, month=10, day=4, hour=2)
pdt3 = datetime(year=2020, month=1, day=3, hour=6)
native_array = [pdt1, pdt2, pdt3]
print(sys.getsizeof(native_array))

ndt2 = np.datetime64("2019-10-04 12:46:00")
ndt3 = np.datetime64("2020-01-03 18:00:32")
numpy_array = np.array([ndt1, ndt2, ndt3], dtype='datetime64')
print(sys.getsizeof(numpy_array))
````

Python's `datetime` objects are high-level abstractions, and storing them in a list introduces overhead due to their object-oriented nature and additional metadata. This becomes noticeable when handling even a small number of `datetime` objects.

On the other hand, NumPy, a library optimized for numerical computations, offers the `np.datetime64` class, which is designed for efficient storage and manipulation of date and time data. NumPy arrays store these objects in a compact, low-level format, making them more memory-efficient when working with larger datasets.

In the example provided, the memory usage of a native Python list containing `datetime` objects is compared to a NumPy array of `datetime64` objects. While the NumPy array demonstrates superior memory efficiency for larger arrays, the benefits may not be as apparent for smaller arrays due to NumPy's memory management overhead.

**In summary**, NumPy is highly advantageous for handling large arrays of datetime data, offering significant memory savings and computational efficiency. However, for smaller arrays, the memory optimization may be negligible.

## Converting between `np.datetime64` and `datetime`

Converting between Python `datetime` and Numpy `datetime64` is typically easy. We saw before that you can initialise a Numpy object from a Python object. This is exactly how we can create Numpy objects out of existing Python objects in one line. To reverse this procedure and get a Python object from a Numpy object, we need to call the `np.astype` function on the `Numpy` object. 

````python
pdt_now = datetime.now() # datetime in Python
print(pdt_now)

ndt_now = np.datetime64(pdt_now) # Convert to Numpy
print(ndt_now)

pdt_fromnp = ndt_now.astype(datetime) # Convert Numpy back to datetime
print(pdt_fromnp)
print(pdt_now == pdt_fromnp)
````

However, there is one important **caveat** to converting between Python and Numpy: Numpy does not respect or preserve **aware** datetimes, meaning datetimes with an attached timezone. Always use naive datetimes in Numpy.

## Using `np.datetime64` in `numpy `

Numpy can generate sequences of datetime objects using the standard Numpy library.

````python
# Use arange to extract every day in February 2020
np.arange("2020-02-01", "2020-03-01", dtype="datetime64")
````

````python
# Using the step argument to return every other day
np.arange("2020-02-01", "2020-03-01", step=2, dtype="datetime64")
````

**What is this `dtype='datetime64[D]'`?**

When comparing memory footprint between Python and Numpy we declared a standard `np.array` object with elements of type `np.datetime64` without explanation. We'll now discuss this further to unpack how Numpy makes sequences of datetime objects manageable. Typically, when considering Numpy arrays we use integer or float numbers and imagine them as big N-dimensional data structures full of a basic (but precise) numbers. Conversely, when looking at datetime objects, we examine how they are anything _but_ basic with many functions and utilities associated with the class. Numpy brings these two features together to allow you to use complex `np.datetime64` objects inside Numpy arrays in **exactly the same fashion as you would for number types**. This is why we can declare a `np.array()` object, specify the element class as `np.datetime64` (using `dtype=np.datetime64`) and then manipulate this data with efficient Numpy operations.

Because `datetime64` behaves just like a number in the Numpy library, we can use functions such as `np.arange()` to generate sequences of datetimes. For `np.arange()` we specify the start datetime (inclusive) and the end datetime (exclusive) as ISO8601 strings. We can also use the `step` argument to extract every _other_ day. To decide on the increment, Numpy will use the smallest specified unit. Since we provided only dates without times, Numpy increased the *day* in each element. We'll discuss how to modify this later but as a hint, the key is in the `[D]` part of the `array.dtype` attribute.

## Resolution and Rescaling

**What does `dtype='datetime64[D]'` mean?** 


Numpy uses a set of **time units** to specify ranges and the **resolution** of a `datetime64` object. 

Specifying the time unit in Numpy functions on `datetime64` objects controls how Numpy handles datetimes manipulation.

The full specification can be found in the Numpy docs [Numpy Datetime Unit Documentation](https://numpy.org/doc/stable/reference/arrays.datetime.html?highlight=datetime#datetime-units)

````python
df = pd.DataFrame({"Code": ["Y", "M", "W", "D", "h", "m", "s", "ms", "..."], 
                   "Time Unit": ["year", "month", "week", "day", "hour", "minute", "second", "millisecond", "smaller units..."]})

df
````

The `[D]` part of the Numpy arrays we just created indicated that those arrays were incrementing in Days. Numpy has a set codes to indicate the unit of time that is being varied or changed. This is different to how Python indexes time units and we will see how this approach can be extremely useful in arithmetic. We show the codes for Year to millisecond but you can specify down to attosecond (one quintillionth of a second) if required. We will also see how these units can be used with raw numbers to specify changes in arithmetic operations. Using a specific unit will change how Numpy manipulates data and can make handling datetime objects far easier than possible in Python alone.

## Examples using Datetime units

````python
# Use the second argument to datetime64 to specify precision of the object
np.datetime64("2020-01-02", "ns")
````

````python
# Specifying hours as the smallest unit - Numpy will increment hours
dt = np.arange("2020-02-01T00", "2020-03-01T00", step=2, dtype="datetime64")
print(dt[:8])
````

We can use these time codes to ensure **consistent** behaviour when handling datetime64 objects. The `np.datetime64` constructor can take a maximum of two arguments with the second argument, a time unit code, specifying the **precision** of the object. We've declared an object with only year, month and day but specified `ns` (nanoseconds) in the second argument. This forces Numpy to fill in the object with values down to this precision. Conversely, if we change `ns` to `Y` (year), then Numpy would only retain the value of Year and drop the smaller datetime details.

We can also look at how these time units work for **varying** datetime objects in the `np.arange()` function. First, we specify hours as the smallest unit and let Numpy guess the unit to increment. Whichever unit that `numpy` decides to vary will impact whether the array spans minutes, hours, days or any other time unit. Here we have specified to select every **other** unit using the `step` argument. The end result is an array over `hours`, increasing by 2 in each element.

````python
# Or use datetime unit code to achieve the same result
dt = np.arange("2020-02-01", "2020-03-01", step=2, dtype="datetime64[h]")
print(dt[:8])
````

````python
# Numpy increments month since we specify [M]
dt = np.arange("2020-02-05", "2020-08-07", dtype="datetime64[M]")
print(dt)
````

The preferred approach to ensuring that Numpy uses hours is by giving the `[h]` code to `dtype`. While these two approaches give the same result, the second strategy using time unit codes removes any ambiguity. For example, if you are importing data formatted with hours, minutes and seconds but you need to increment hours between each imported item -- it would be far easier to use `dtype=datetime64[h]` in `np.arange()` than to splice each date so that Numpy guesses to increment hours. We can also increment larger time units such as month using `[M]`, but it is important to remember that Numpy will "cut off" the smaller time units. Here, we see that Numpy cuts off days when asking for the months between February and August. 

## Useful functions

* Numpy includes a `timedelta64` class, similar to the Python `datetime.timedelta`, for time **intervals** and **differences** between datetime objects. 
* We can use the **time units** to specify the length of the `timedelta64`.
* Numpy supports only one unit in a `timedelta64` object.

````python
np.timedelta64(14, "m") # 14 minutes
````

````python
np.timedelta64(366, "ms") # 366 milliseconds
````

Numpy also has a class for time intervals similar to Python. Before, the Python `timedelta` class accepted arguments for each time unit. However, for Numpy we need to use the codes we just saw to specify the duration of the delta. In addition, Numpy supports only one unit for a `timedelta64` object. The same types of units from the ISO8601 format are accepted so here we have declared `timedelta64` objects in minutes (`m`) and milliseconds (`ms`). Note that these units are case sensitive - a capital `M` will give you a `timedelta64` in months not minutes!

````python
np.timedelta64(18, "W") # 18 weeks
````

````python
np.timedelta64(7, 'h') + np.timedelta64(7, 'm') # 7 hours and 7 minutes
````

Similar to Python, the `numpy` class supports weeks as `W`. In our last example, we desired an interval of 7 hours and 7 minutes. To support this, all units are converted to the smallest unit required. Numpy stores the output as 427 minutes, which is equivalent to the desired duration in a single unit. 

Convert a `np.timedelta64` interval into a new interval by **creating a new object** with the desired time unit.

````python
dt = np.timedelta64(7, 'h') + np.timedelta64(7, 'm') + np.timedelta64(366, "ms")
np.timedelta64(dt, "s") # Create a new timedelta64 in seconds
````

Or extract the whole seconds in a `timedelta64` object using `np.astype()`.

````python
dt.astype("timedelta64[s]") # Convert to seconds but drop the 366ms
````

To change the unit of a `timedelta64` object, we can create a new object with the desired unit code. We can also use casting with `np.astype()` to convert timedelta objects into different units. These two approaches are mostly similar but with a subtle distinction. Creating a new object will check if the conversion is valid (more on this shortly) whereas using `np.astype()` will convert the duration using default values. These two will give you the same result until you need to convert time units with inconsistent lengths.

Here we create a timedelta containing hours, minutes and milliseconds. We want to determine how many seconds are in this interval. The fastest approach is to cast the object to `timedelta64[s]`, which transforms the object into seconds. However, this will drop the units smaller than a second meaning we lose 366 milliseconds.

Or we could extract the total seconds (including fractional seconds) by dividing by a 1 second `timedelta64`.

````python
seconds = dt / np.timedelta64(1, 's')
print(seconds)
````

Numpy has a special case for datetime errors called `NaT` (*not a time*). This is a variant of `NaN` (*not a number*).

````python
np.datetime64("nat")
np.timedelta64("nat")
````

In addition, Numpy stores a special case for when there is an error or the output isn't a valid datetime. This `NaT` value is good to know about because it usually means something has gone wrong in your system!

## Arithmetic with `np.datetime64`

Operations between a `datetime64` and a `timedelta64` **return** a `datetime64`.

````python
ndt1 = np.datetime64("2018-10-03 13:00:00")
ndt2 = np.datetime64("2019-10-04 12:46:00")
delta1 = np.timedelta64(14, "D")

print(ndt1 - delta1)
print(ndt2 + delta1)
````

Arithmetic using Numpy objects is very similar to using Python `datetime` objects. The same rules apply to what we saw in our previous lesson. Just keep in mind whether your desired outcome is a **datetime** or a **timedelta** to check you are using the correct operation. Some arithmetic will return a time difference and others give you a fixed point in time. Arithmetic between a `datetime64` and a `timedelta64` will return a new `datetime64`.

**Subtracting** a `datetime64` from another `datetime64` will return the elapsed time between these two objects as a `timedelta64` object.

````python
ndt2 - ndt1
````

You **cannot** add two `datetime64` objects.

Always use **time unit** codes to ensure your arithmetic changes a `datetime64` correctly.

````python
ndt1 + np.timedelta64(7, "D") # Add 7 days to a datetime
````

Since **Months** and **Years** have variable lengths - you cannot convert a `timedelta64` in either unit into days units. This is to ensure **consistency** in Numpy.

````python
d1 = np.datetime64('2026-02')          # month precision: <M8[M]
d2 = d1 + np.timedelta64(1, 'M')       # 2026-03

(d2.astype('datetime64[D]') - d1.astype('datetime64[D]')) / np.timedelta64(1, 'D')
# 28.0
````

Understanding the time unit codes are the key to mastering `datetime64` in Numpy. We previously mentioned that there is a subtle distinction when converting some `timedelta64` objects with variable length into other units. Specifically, some units such as month and year aren't always the same size. Therefore, Numpy will prohibit you to convert a 1 month interval into days by default because we can't know whether this should be 28 days or 31 days. Similarly, a 1 year interval could be 365 days or 366 days. These safety checks only exist **when creating a new `np.timedelta64` object** and **not** when using `np.astype()`. The `astype()` behaviour will use default values in conversion so the result here would be 30. This _might_ be correct, but not always because months aren't always 30 days. Keep this in mind when converting between `timedelta64` units.

## Using `datetime64` for working days
Another useful feature of `np.datetime64` is built in functionality for detecting and handling **business days**. 

The `is_busday()` function return `True` if a given `np.datetime64` is in the Monday-Friday working week.

````python
np.is_busday(np.datetime64("2020-08-12"))
````

`busday_count()` will return the number of business days between two dates.

````python
ndt1 = np.datetime64("2020-09-20")
ndt2 = np.datetime64("2020-12-15")

np.busday_count(ndt1, ndt2)
````

Numpy includes some built in functions for handling business days and working weeks inside `np.datetime64`. While this will not always be useful, the time spent in understanding these functions is far less than the time reimplementing such functionality if you didn't know this existed. For example, this can be useful if you need to calculate shipping estimates for logistics pipelines or calculate productivity hours in given weeks. 

The `is_busday()` function is the simplest utility which only returns if a given date is a business day or not. The `busday_count()` functionality will return the number of business days between two dates. This is equivalent to summing all the days in this interval for which `is_busday()` returns `True`. Remember, the end date is always excluded. 

The `busday_offset()` function will calculate an offset of `k` business days from a given date.

````python
ndt1 = np.datetime64("2020-08-12")
k=2
np.busday_offset(ndt1, k)
````

Hence if you ask for an offset of 1 from a Friday, then the function will return the following Monday. 

This function assumes the input date is **also** a business day. If this is not true then the input day is **adjusted** using the `roll` argument.
- `roll='forward'`: adjust to first valid day in the future.
- `roll='backward'`: adjust to first valid day in the past.
- `roll='nat'`: return Not A Number.
- `roll='raise'`: raise exception instead of adjusting (default).

````python
ndt2 = np.datetime64("2020-08-08") # Saturday 8th August 2020
np.busday_offset(ndt2, k, roll="forward")
````

Here, we ask for 2 business days from `ndt2` but this object represents a Saturday. By specifying `roll=forward`, the date is adjusted to the following Monday and only then is the offset applied. We can also specify which days are considered business days using the `weekmask` argument.

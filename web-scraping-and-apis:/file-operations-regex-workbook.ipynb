## The vocabulary

Regular expressions are a way to define strings that match a specific pattern. There is a whole language to be learned here and below you will find some of the most useful values.

    \d      Any Digit
    \D      Any Non-digit character
    .       Any Character
    \.      Period
    [abc]   Only a, b, or c
    [^abc]  Not a, b, nor c
    [a-z]   Characters a to z
    [0-9]   Numbers 0 to 9
    \w      Any Alphanumeric character
    \W      Any Non-alphanumeric character
    {m}     m Repetitions
    {m,n}   m to n Repetitions
    *       Zero or more repetitions
    +       One or more repetitions
    ?       Optional character
    \s      Any Whitespace
    \S      Any Non-whitespace character
    ^…$     Starts and ends
    (ab|de) Matches ab or de

[source](https://regexone.com/references/python)

A fantastic way of learning/practicing regex is [regex101](https://regex101.com/)

# Match patterns

Firstly, try to understand the regular expression below:

````python
r'GET /\w+'
````

This regular expression pattern looks for the word **`GET`**, followed by a **space character** ` `, a **forward slash** `/`, and then one or more **alphanumeric characters** (letters, digits, or underscore). This pattern is commonly used to search for **simple URLs** requested with a `GET` method in logs or text.

Although not always required, it's a good practice to prefix regex patterns with **`r''`** (raw strings). This ensures that backslashes `\` are treated **literally** and not as Python escape characters. For example, `\n` is a newline in a normal string but just a backslash and "n" in a raw string.

# Match dates

Another example of a simple regular expression would be to search for a date. Imagine we wanted to capture all date instances that followed the convention 01/Jul/1995. This pattern would look like:

````python
r'\d{2}/[a-zA-Z]{3}/\d{4}'
````

Pattern Breakdown:
- `\d{2}` — Matches **exactly two digits**, e.g. `01`, `23`  
- `/` — Matches a **literal forward slash**  
- `[a-zA-Z]{3}` — Matches **exactly three letters**, either uppercase or lowercase, e.g. `Jul`, `DEC`  
- `/` — Matches another **literal forward slash**  
- `\d{4}` — Matches **exactly four digits**, e.g. `1995`, `2023`

With our log we know exactly the number of letters and numbers expected. With regex, you can use `{n}` after a pattern to specify that you expect **n** instances. For example `\d{4}` will match exactly 4 digits.

## ✅ Question 3: `search()`

Python supports regular expressions through a built-in library called `re`. Start by importing the `re` library and using 
the `.search()` function, which takes a regex pattern and a string to be searched. 

- Search `log_file` for `GET` requests (the regex specified under the vocublary list above can be used) `r'GET /\w+'`

- then assign the `re.Match object` which is returned by `re.search()`, to the variable `match`

See below code syntax for some guidance:

```python
match = re.search(<regex_pattern>, <data>)
```

````python
import re

# Add your code below
pattern = r'GET /\w+'
match = re.search(pattern, log_file)
match
````

Note that the search function will only return the first match, so it is mostly useful to see whether your regular expression does at least have one hit. If there is no hit, it simply returns `None`. If it does find a hit, it returns the hit as a `MatchObject`.

We can use the `.group()` method of the `MatchObject` to extract the actual string that was matched:

````python
match.group()
````

We can use the `.start()` method of the `MatchObject` to extract the starting position of the string that was matched:

````python
match.start()
````

We can use the `.end()` method of the `MatchObject` to extract the end position of the string that was matched:

````python
match.end()
````

### Creating a message from `MatchObject` properties

Similarly to the `.group()` method which extracts the actual string which was matched, you can use the `.start()` and `.end()` methods to collect the position of where the regular expression matched.

`print` the `start` and `end` positions formatted nicely with the actual match. Use an f-string to put together the `.group()`, `.start()`, `.end()` into a sentence as a single string, assigned to `message`:

````python
# return a single string containing .group(), .start() and .end() of the match
message = f'Found "{match.group()}" at position: {match.start()}-{match.end()}'
message
````

## ✅ Question 4: Finding and Replacing IP Addresses

Your task is to **identify and replace all IPv4 host IP addresses** in the `log_file` with the string **`'unknown domain'`**.

This task will be completed in **two steps**:

1. **Identify** all IP addresses in the file  
2. **Replace** each IP address with `'unknown domain'`

To help with this, you've been provided the following regular expression pattern:

```python
r'[0-9]*\.[0-9]*\.[0-9]*\.[0-9]*'
```

This pattern is designed to match IPv4 addresses such as `192.168.0.1`. Here's a quick breakdown of how it works:

- `[0-9]*` – Matches **zero or more digits**, representing each segment of the IP address  
- `\.` – Matches a **literal dot** (`.`), which separates each segment  
- The pattern is repeated **four times** to match all four parts of a typical IPv4 address

---

#### Step 1: Finding IP Addresses

Now find all IP addresses in the `log_file`, you can use the `re.findall()` function.

See below code syntax for some guidance:

```python
pattern = r'[0-9]*\.[0-9]*\.[0-9]*\.[0-9]*'

re.findall(<regex_pattern>, <data>)
```

````python
# Step 1: Finding IP Addresses
# Add your code below

re.findall(pattern, log_file)
````

#### Step 2: Replacing IP Addresses and Saving the File

Now that we've identified the IP addresses, let's move on to replacing them using `re.sub()` function.

We'll use the same regular expression pattern to substitute every IPv4 address in the `log_file` with the string **`'unknown domain'`**.

Finally, we'll **save the updated content** to a new file called `log_file_edited.log`

````python
# Step 2: Replacing IP Addresses and Saving the File

pattern = r'[0-9]*\.[0-9]*\.[0-9]*\.[0-9]*'
replacement = r'unknown domain'

with open('data/log_file_edited.log', 'w') as f:
    f.write(re.sub(pattern, replacement, log_file))
````

````python
# Read and Print the Edited File for verification

with open('data/log_file_edited.log', 'r') as f:
    content = f.read()
    print(content)
````

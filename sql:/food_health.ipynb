# ☑️ Introduction to the Food Health dataset

This dataset has been taken from an open data platform (found __[here](https://data.sfgov.org/)__) hosted by the city of San Francisco. It is an sqlite file which is a database created using SQLite. SQLite is a lightweight verion of SQL which features similar syntax to the vast majority of SQL flavours. It provides a fast way of accessing databases and storing databases locally. 

The data includes restaurant scores produced by the San Francisco health department up to and including the year 2016. These scores are generated by a health inspector who bases them upon the violations observed, these violations fall into three categories: High risk, Moderate risk or Low risk.

## Part 3: Subqueries and joins

**11. From the businesses table, find all owners that own 5 or more restaurants. Then  find the 10 most popular locations for  restaurants (using postal_code) amongst the owners who have 5 restaurants or more. The final result should return the 10 most popular areas  (postal_code) and the frequency with which they appear in our group of owners who own five restaurants or more. The result should have two columns (postal_code and frequency) and should be presented in descending order of frequency.**

Please note you have been provided with the code for this question to carry out the necessary data manipulation and reporting work. Simply run the code cell to produce the desired results.

````python
most_popular_post_codes_query = """
SELECT postal_code, COUNT(*) AS frequency
FROM businesses
WHERE owner_name IN (
SELECT DISTINCT(owner_name)
FROM (SELECT owner_name, COUNT(business_id) AS num_restaurants
    FROM businesses
    GROUP BY owner_name
    ORDER BY num_restaurants DESC) AS restaurant_owners
WHERE restaurant_owners.num_restaurants >= 5)
GROUP BY postal_code
ORDER BY frequency DESC
LIMIT 10;
"""

most_popular_post_codes_result = pd.read_sql(most_popular_post_codes_query, conn)
most_popular_post_codes_result
````

**12. Now it might be interesting to look at some statistics. For all the restaurants in the "94103" post code let's calculate the minimum score (as "min_score"), average score (as "avg_score") and maximum score (as "max_score"). The average score should be rounded to one decimal and you should only consider restaurants that have undergone inspection (so the score is NOT NULL).**

- It is recommended to utilize SQL functions such as `MIN()`, `AVG()`, `ROUND()`, and `MAX()`
- Ensure to include SQL `NOT NULL` when specifying criteria for the `Score` column, and also consider filtering by `postal_code`=`94103`
- Additionally, an `INNER JOIN` can be used to merge the `businesses` and `inspections` tables based on the common field `business_id`

See below code syntax for some guidance:
```SQL
SELECT <column_names>
FROM <table1>
INNER JOIN <table2>
ON table1.column_name = table2.column_name
WHERE <condition1> AND <condition2>;
```

````python
#add your code below
min_avg_max_score_query = """

SELECT 
    MIN(score) AS min_score,
    ROUND(AVG(score), 1) AS avg_score,
    MAX(score) AS max_score
FROM businesses
INNER JOIN inspections
ON businesses.business_id = inspections.business_id
WHERE inspections.score IS NOT NULL
  AND businesses.postal_code = '94103';


"""
min_avg_max_score_result = pd.read_sql(min_avg_max_score_query, conn)
min_avg_max_score_result

````

**13. Now we can get a bit more serious and look at how many times restaurants in the "94103" post code have committed health violations. We can then group them based on their risk category so we know which restaurants to avoid. The output should have two columns (risk_category, frequency) and be sorted in descending order by frequency**

- Consider using SQL `COUNT(*)` function
- Additionally, an `INNER JOIN` can be used to merge the `businesses` and `violations` tables based on the common field `business_id`
- Consider filtering by `postal_code`=`94103`


See below code syntax for some guidance:
```SQL
SELECT <column_names>
FROM <table1>
INNER JOIN <table2>
ON table1.column_name = table2.column_name
WHERE <condition>
GROUP BY <column_name>
ORDER BY <column_name> DESC;
```

````python
#add your code below
market_street_health_violations_query = """

SELECT 
    violations.risk_category, 
    COUNT(*) AS frequency
FROM businesses
INNER JOIN violations
ON businesses.business_id = violations.business_id
WHERE businesses.postal_code = '94103'
GROUP BY violations.risk_category
ORDER BY frequency DESC;


"""
market_street_health_violations_result = pd.read_sql(market_street_health_violations_query, conn)
market_street_health_violations_result

````

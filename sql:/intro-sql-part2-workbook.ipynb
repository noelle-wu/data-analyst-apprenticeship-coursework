# ☑️ SQL Statements

**Q1)** Refer to the `airlines` table. How many airlines start with a number between `0` and `9`?
- You can use `COUNT(*)` to count the number of rows returned

Please note you have been provided with the code for this question to carry out the necessary analysis work. Simply run the code cell to produce the desired results.

````python
query = """
SELECT COUNT(*) 
FROM airlines 
WHERE SUBSTR(name, 1, 1) BETWEEN '0' AND '9';
"""
df = pd.read_sql_query(query, conn)
df
````

**Q2)** Refer to the `country` column in `airlines` table. How many countries have at least one airline?

- Consider `IS NOT NULL` keyword to filter out any NULL values in `country` column

- Use DISTINCT() function to extract distinct values and use COUNT() function to calculate the number of distinct values in `country` column

See below code syntax for some guidance:
```SQL
SELECT COUNT(DISTINCT(column_name))
FROM airlines 
WHERE <column_name> IS NOT NULL;
```

````python
#add your code below
query = """
SELECT COUNT(DISTINCT(country))
FROM airlines 
WHERE country IS NOT NULL;
"""

df = pd.read_sql_query(query, conn)
df
````

**Q3)** Refer to the `altitude` column in the `airports` table. Find the **five** airports with the highest `altitude`.  

Please note that the `altitude` column is stored as a **text** data type, which prevents the `ORDER BY` operation from being performed correctly.

- Refer to the `ORDER BY` keyword, and use the **CAST()** function to convert the `altitude` column to the `INT` data type.  
  - **Example**: `CAST(altitude AS INT)`  
- The values should now be ordered correctly.

See below code syntax for some guidance:

```SQL
SELECT name, country, altitude
FROM airports
ORDER BY CAST(<column_name>) DESC
LIMIT 5;
```

````python
#add your code below
query = """

SELECT name, country, altitude
FROM airports
ORDER BY CAST(altitude AS FLOAT) DESC
LIMIT 5;
"""

df = pd.read_sql_query(query, conn)
df
````

**Q4)** Refer to the `name` and `altitude` columns in `airports` table. Further filter your data to get all airports from `United Kingdom`.

See below code syntax for some guidance:
```SQL
SELECT name, altitude
FROM airports
WHERE country='United Kingdom';
```
Now use CASE() statement to create a new calculated column called `altitude2` to reflect below conditions and results:
- Return "Low" if altitude lower than 100m
- Otherwise return "Medium" if altitude lower than 500m
- Otherwise return "High"

See below code syntax for some guidance:
```SQL
CASE
    WHEN CAST(altitude AS INT) < 100 THEN 'Low'
    WHEN CAST(altitude AS INT) < 500 THEN 'Medium'
    ELSE 'High'
END AS altitude2
```

````python
#add your code below
query = """
SELECT name, altitude,
CASE
    WHEN CAST(altitude AS INT) < 100 THEN 'Low'
    WHEN CAST(altitude AS INT) < 500 THEN 'Medium'
    ELSE 'High'
END AS altitude2
FROM airports
WHERE country='United Kingdom';

"""

df = pd.read_sql_query(query, conn)
df
````

This section illustrates how a user may try to use a function in an unintended way. To overcome such situation you have been given the below code for the function `error_catching`. As you can see this function will display **'Only numbers should be used. Please try again.'** if any of the parameters is a string.

```python
def error_catching(a, b, c):
    if (type(a) == str) or (type(b) == str) or (type(c) == str):
        result = 'Only numbers should be used. Please try again.'
    else:
        result = (a + b) * c
    
    return result
```

```python
print(error_catching(5, 6, 'seven'))
```

**Q2)** Now create a function to calculate how many values in a `list` are duplicated. Recall that `len()` will give the length of a `list` or `set`

- Create a function called `duplicate_count` which should return the number of duplicate values in a list:
- Call the `unique()` function within the `duplicate_count()` function
- Also, remember to avoid using a parameter name which has been used as a variable name elsewhere

```python
#add your code below
def duplicate_count(dupe_list):
    total = len(dupe_list)
    unique_count = len(unique(dupe_list))
    dupes = total - unique_count
    return dupes

```

Test your code using `names_list`.

```python
duplicate_count(names_list)
```

**Example3:** You have been given the below code for the function called `names_dict`, which takes a `list` and returns a `dictionary` where:
- there is a `key` for each unique first character found in the list   
- the `values` are a list of the entries which begin with the character used for the `key`
- the lists of entries are ordered alphabetically

We will build up `names_dict` function step-wise below. The below function takes a list of strings (eg, names) and returns a list of unique, first characters of those names, `keys`.  Test it using the list provided.

```python
def names_dict(names):
    first_chars = []
    
    for name in names:
        first_chars.append(name[0])
    
    keys = set(first_chars)
    keys = sorted(keys)
    
    return keys
```

**Q4)** Copy your code for `names_dict` from above to the cell below and expand it:
- Creating a dictionary where the keys are the letters in the list `keys`, and the values are lists of names that begin with that letter. 
- Again, test your code using `test_list`.

**See below example of what the expected output should look like:** 
<code>
{'A': ['Alan', 'Adrian'],
 'C': ['Charlene', 'Carol', 'Clive', 'Colin'],
 'D': ['Donald'],
 'G': ['Grace', 'Gillian'],
 'J': ['Jenna'],
 'L': ['Lindsey', 'Lindsey', 'Lindsey'],
 'R': ['Ricky']}
</code>

```python
#add your code below

def names_dict(names):
    first_chars = []
    
    for name in names:
        first_chars.append(name[0])
        
    keys = set(first_chars)
    keys = sorted(keys)
    
    new_dict = dict()
    
    for key in keys:
        name_list = []
        for name in names:
            if name[0] == key:
                name_list.append(name)
                
        new_dict[key] = name_list
        
    return new_dict
```
